## Task 2: Unemployment Analysis with Python

### Project Objective:
- Clean and preprocess unemployment dataset tracking rates across Indian states and geographical zones.
- Investigate the impact of COVID-19 pandemic restrictions and lockdown on unemployment rates.
- Identify seasonal trends, state-wise disparities, and provide actionable economic insights.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style="whitegrid")
```

### 1. Load Data & Preprocessing

```python
df = pd.read_csv('Unemployment_Rate_upto_11_2020.csv')
df.columns = df.columns.str.strip()
df.rename(columns={
    'Region': 'State',
    'Region.1': 'Zone',
    'Estimated Unemployment Rate (%)': 'Unemployment_Rate',
    'Estimated Employed': 'Employed',
    'Estimated Labour Participation Rate (%)': 'Labour_Participation_Rate'
}, inplace=True)

df['Date'] = pd.to_datetime(df['Date'].str.strip(), format='%d-%m-%Y')
df['Month'] = df['Date'].dt.month_name()
df['Year'] = df['Date'].dt.year
df.head()
```

```python
print('Data Shape:', df.shape)
print('\nMissing values:\n', df.isnull().sum())
df[['Unemployment_Rate', 'Employed', 'Labour_Participation_Rate']].describe().T
```

### 2. Unemployment Trends: COVID-19 Impact Analysis

```python
monthly_trend = df.groupby('Date')['Unemployment_Rate'].mean().reset_index()

plt.figure(figsize=(10, 5))
plt.plot(monthly_trend['Date'], monthly_trend['Unemployment_Rate'], marker='o', color='firebrick', linewidth=2.5)
plt.axvline(pd.to_datetime('2020-03-25'), color='black', linestyle='--', label='COVID-19 Lockdown (March 2020)')
plt.title('Monthly Average Unemployment Rate in India (2020) - COVID-19 Impact', fontsize=13, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.legend()
plt.show()
```

### 3. State-wise Unemployment Analysis

```python
state_avg = df.groupby('State')['Unemployment_Rate'].mean().sort_values(ascending=False).reset_index()

plt.figure(figsize=(12, 8))
sns.barplot(data=state_avg, x='Unemployment_Rate', y='State', hue='State', palette='rocket', legend=False)
plt.title('Average Unemployment Rate by State (2020)', fontsize=14, fontweight='bold')
plt.xlabel('Average Unemployment Rate (%)')
plt.ylabel('State')
plt.show()
```

### 4. Zone-wise Distribution & Disparities

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x='Zone', y='Unemployment_Rate', hue='Zone', palette='Set2', legend=False)
plt.title('Unemployment Rate Distribution across Geographic Zones', fontsize=13, fontweight='bold')
plt.xlabel('Geographic Zone')
plt.ylabel('Unemployment Rate (%)')
plt.show()
```

### 5. Pre-Lockdown vs. Lockdown Period Comparison

```python
df['Period'] = np.where(df['Date'] < '2020-04-01', 'Pre-Lockdown (Jan-Mar)', 'Post/During Lockdown (Apr-Oct)')
period_comp = df.groupby(['Zone', 'Period'])['Unemployment_Rate'].mean().unstack()

period_comp.plot(kind='bar', figsize=(10, 6), colormap='viridis', edgecolor='black')
plt.title('Unemployment Rate: Pre-Lockdown vs. Lockdown across Zones', fontsize=13, fontweight='bold')
plt.ylabel('Average Unemployment Rate (%)')
plt.xlabel('Zone')
plt.xticks(rotation=0)
plt.show()
```

### 6. Correlation Heatmap of Economic Indicators

```python
numeric_cols = ['Unemployment_Rate', 'Employed', 'Labour_Participation_Rate', 'longitude', 'latitude']
plt.figure(figsize=(8, 6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix of Economic Indicators', fontsize=13, fontweight='bold')
plt.show()
```

### 7. Key Findings & Policy Insights
- **Pre-Lockdown Average:** ~9.76%
- **Peak Lockdown Surge:** Reached 23.24% in May 2020 (a surge of +8.99% overall during peak restrictions).
- **Most Impacted States:** Haryana (27.48%), Tripura (25.05%), Jharkhand (19.54%), Bihar (19.47%), Delhi (18.41%).
- **Least Impacted States:** Meghalaya (3.87%), Assam (4.86%), Gujarat (6.38%), Odisha (6.46%).
